#Queue Pattern (Pattern #15)

Queue is very simple if you visualize correctly.

Step 1: Real-life example (best way to understand)

Think of people standing in a line:

Person A → Person B → Person C → Person D

Who enters first? → A
Who leaves first? → A

This is called:

FIFO = First In First Out

This is Queue.

Step 2: Visual representation

Queue looks like this:

Front → [10, 20, 30, 40] ← Rear

Front = first element
Rear = last element

Remove from Front
Add to Rear

Step 3: Queue operations (only 2 important)
Operation 1: Enqueue (add element)

Add element at rear

Example:

[10, 20, 30]
enqueue(40)

Result:

[10, 20, 30, 40]
Operation 2: Dequeue (remove element)

Remove element from front

Example:

[10, 20, 30, 40]
dequeue()

Result:

[20, 30, 40]

10 removed.

Step 4: Queue in Python (using list)
queue = []

# enqueue
queue.append(10)
queue.append(20)
queue.append(30)

print(queue)

Output:

[10, 20, 30]
Step 5: Dequeue in Python
queue.pop(0)

Example:

queue = [10, 20, 30]

queue.pop(0)

print(queue)

Output:

[20, 30]
Step 6: Full Queue example
queue = []

# enqueue
queue.append(10)
queue.append(20)
queue.append(30)

print("Queue:", queue)

# dequeue
queue.pop(0)

print("After dequeue:", queue)

Output:

Queue: [10, 20, 30]
After dequeue: [20, 30]
Step 7: Visualization
enqueue(10) → [10]
enqueue(20) → [10, 20]
enqueue(30) → [10, 20, 30]

dequeue() → remove 10 → [20, 30]
Step 8: Queue core pattern template
queue = []

# enqueue
queue.append(x)

# dequeue
queue.pop(0)
Step 9: Pattern recognition keywords

If problem says:

queue

FIFO

first in first out

process in order

→ Use Queue pattern

Step 10: Time complexity

Enqueue:

O(1)

Dequeue:

O(n) using list

Later we use deque for O(1).

```python
queue = []

queue.append(10)
queue.append(20)
queue.append(30)

print(queue)

queue.pop(0)
```
