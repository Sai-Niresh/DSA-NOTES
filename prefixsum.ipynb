Step 1: Understand the problem (normal way)

Example:

arr = [2, 4, 6, 8]

Find sum from index 1 to 3:

4 + 6 + 8 = 18

Normal way:

sum(arr[1:4])   # O(n)

This is slow if many queries exist.

Step 2: Prefix Sum Idea (core concept)

Prefix sum means:

prefix[i] = sum of elements from index 0 to i

Example:

arr    = [2, 4, 6, 8]
prefix = [2, 6, 12, 20]

How?

prefix[0] = 2
prefix[1] = 2+4 = 6
prefix[2] = 6+6 = 12
prefix[3] = 12+8 = 20
Step 3: Visualization
index:   0   1   2   3
arr:     2   4   6   8
prefix:  2   6  12  20
Step 4: Range sum using prefix sum

Formula:

sum(L to R) = prefix[R] - prefix[L-1]

Example: sum(1 to 3)

prefix[3] - prefix[0]
20 - 2 = 18


```python
arr = [2, 4, 6, 8]
prefix = [0] * len(arr)

prefix[0] = arr[0]

for i in range(1, len(arr)):
    prefix[i] = prefix[i-1] + arr[i]

print(prefix)    
```
