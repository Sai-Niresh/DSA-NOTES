```python
'''
arr = [1, 3, 5, 7, 9]
target = 10
Find if any two numbers add up to 10.
Output: True
'''
def sum_two_num(arr, target):
    left = 0
    right = len(arr)-1
    while left < right:
        sum = arr[left] + arr[right]
        if sum == target:
            return True 
        elif sum < target:
            left += 1
        else:
            right -= 1
    return False
print(sum_two_num([1, 3, 5, 7, 9], 10))
```

```python
'''
arr = [1, 2, 3, 4, 2]
Find if array has any duplicate.
Output: True
'''

def find_duplicate(arr):
    freq = {}
    for num in arr:
        if num in freq:
            freq[num] += 1
        else:
            freq[num] =1
        
    for num in arr:
        if freq[num] > 1:
         return True
print(find_duplicate([1, 2, 3, 4, 2]))
```
