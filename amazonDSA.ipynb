arr = [5, 3, 8, 1, 9, 2]
Find the largest number.
Output: 9

```python
def largest_num(arr):
    max_val = arr[0]
    for num in arr:
        if num > max_val :
            max_val = num
    return max_val 
arr = [5,3,8,1,9,2]
print(largest_num(arr))
```

s = "hello"
Check if it is a palindrome.
Output: False

```python
def palindrome(s):
    if s == s[::-1]:
        return True
    else:
        return False
s = "racecar"
print(palindrome(s))
```

```python


def find_duplicate(arr):
    seen = set()
    for num in arr:
        if num in seen:
            return True
        else:
            seen.add(num)
    return False
arr = [3, 1, 4, 1, 5, 9, 2]
print(find_duplicate([3, 1, 4, 1, 5, 9, 2]))
```
