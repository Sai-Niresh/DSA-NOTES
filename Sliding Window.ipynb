#Sliding Window Pattern (from scratch)

This is one of the most important patterns for interviews.

📌 First understand the problem type

Example problem:

Find the sum of every 3 consecutive elements

arr = [2, 1, 5, 1, 3, 2]
k = 3


We need sums of windows of size 3:

[2,1,5] → sum = 8
[1,5,1] → sum = 7
[5,1,3] → sum = 9
[1,3,2] → sum = 6

Sliding Window method (fast way)

Idea:
Instead of recalculating sum, reuse previous sum.

Step-by-step idea:

Window moves like this:

[2,1,5] → sum = 8
remove 2, add 1 → sum = 7
remove 1, add 3 → sum = 9
remove 5, add 2 → sum = 6


```python
arr = [2, 1, 5, 1, 3, 2]
k = 3

window_sum = sum(arr[:k])
print(window_sum)

for i in range(k, len(arr)):
    window_sum = window_sum - arr[i - k] + arr[i]
    print(window_sum)
```

#Sliding Window Problem 1: Maximum sum of subarray of size k
Problem

Find the maximum sum of any subarray of size k

```python
arr = [2, 1, 5, 1, 3, 2]
k = 3

window_sum = sum(arr[:k])
max_sum = window_sum

for i in range(k, len(arr)):
    window_sum += arr[i] - arr[i - k]
    if window_sum > max_sum:
        max_sum = window_sum
print(max_sum)                
```

#🧠 Sliding Window Problem 2: Minimum sum of subarray of size k
Problem

Find the minimum sum of any subarray of size k

```python
arr = [2, 1, 5, 1, 3, 2]
k = 3
window_sum = sum(arr[:k])
min_sum = window_sum

for i in range(k, len(arr)):
    window_sum = window_sum - arr[i -k] + arr[i]
    if  window_sum < min_sum:
        min_sum = window_sum
print(min_sum)
```

Sliding Window Problem 3: Count subarrays with sum > target
Problem

Count how many subarrays of size k have sum greater than target

```python
arr = [2, 1, 5, 1, 3, 2]
k = 3
target = 7
count = 0

window_sum = sum(arr[:k])
if window_sum > target:
    count += 1

for i in range(k, len(arr)):
    window_sum = window_sum - arr[i-k] + arr[i]
    if window_sum > target:
        count += 1
print(count)
```
