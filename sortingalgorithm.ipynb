📌 What is Sorting?

Sorting means arranging elements in a specific order.

Most common:

Ascending → small to large

Descending → large to small

Example:

[4, 2, 1, 3] → [1, 2, 3, 4]

🧠 Why Sorting is Important?

Sorting helps in:

Binary search

Removing duplicates

Finding min / max easily

Many interview problems

🔹 SORTING ALGORITHM 1: BUBBLE SORT

We start with Bubble Sort because:

Easy to understand

Builds comparison logic

Uses loops you already know

🫧 What is Bubble Sort?

The largest element “bubbles” to the end step by step.

🔹 Core Idea (Very Simple)

Compare neighbor elements

If left > right → swap

Repeat multiple times

🧩 Example
arr = [5, 3, 4, 1]

Pass 1:
5 > 3 → swap → [3, 5, 4, 1]
5 > 4 → swap → [3, 4, 5, 1]
5 > 1 → swap → [3, 4, 1, 5]


Largest number 5 is now at the end.

```python
arr = [5, 3, 4, 1]

n = len(arr)

for i in range(n):
    for j in range(n - 1 ):
        if arr[j] > arr[j + 1]:
            arr[j], arr[j + 1] = arr[j + 1], arr[j]
print(arr)            
```

# SELECTION SORT (FROM SCRATCH)
📌 What is Selection Sort?

Selection Sort selects the smallest element and puts it in the correct position.

Think like this:

Find the smallest element

Put it at the beginning

Repeat for the rest of the array

🧠 How Selection Sort THINKS (Very Important)

Bubble Sort:

Pushes largest to the end

Selection Sort:

Pulls smallest to the front

```python
arr = [4, 3, 1, 3]

n = len(arr)

for i in range(n):
    min_index = i
    for j in range(i + 1, n):
        if arr[j] < arr[min_index]:
            min_index = j 
    arr[i], arr[min_index] = arr[min_index], arr[i]
print(arr)
```

#NEXT IN SORTING (CONTINUATION)
🔹 1️⃣ Optimized Bubble Sort (Very Important)
📌 Why optimize Bubble Sort?

In normal Bubble Sort:

Even if the array is already sorted

We still keep looping ❌

Optimized Bubble Sort:

Stops early if no swaps happen

Saves time

Shows good problem-solving thinking

```python
arr = [1, 2, 3, 4]
n = len(arr)

for i in range(n):
    swapped = False

    for j in range(n - 1, i):
        if arr[j] < arr[j - 1]:
            arr[j], arr[j - 1] = arr[j - 1], arr[j]
            swapped = True
    if not swapped:
        break
print(arr)
```

```python
arr = [6, 3, 8, 2]
def merge_sort(arr):
    if len(arr) <= 1:
        return arr

    mid = len(arr) // 2
    left = merge_sort(arr[:mid])
    right = merge_sort(arr[mid:])

    return merge(left, right)


def merge(left, right):
    result = []
    i = j = 0

    while i < len(left) and j < len(right):
        if left[i] < right[j]:
            result.append(left[i])
            i += 1
        else:
            result.append(right[j])
            j += 1

    result.extend(left[i:])
    result.extend(right[j:])
    return result


arr = [6, 3, 8, 2]
sorted_arr = merge_sort(arr)
print(sorted_arr)
```

Idea of Insertion Sort

Think about sorting playing cards in your hand.

When you pick a new card:

you compare it with previous cards

you insert it in the correct position.

That is exactly what Insertion Sort does.

```python
arr = [5, 3, 4, 1]

for i in range(1, len(arr)):
    key = arr[i]
    j = i - 1

    while j>= 0 and arr[j] > key:
        arr[j + 1] = arr[j]
        j -=1

    arr[j + 1] = key

print(arr)
```

Quick Sort — Idea

Quick Sort follows Divide and Conquer.

Steps:

Pick a pivot element

Place pivot in correct position

Put smaller elements on left

Put larger elements on right

Recursively sort left and right parts

```python

def quick_sort(arr):
    if len(arr) <= 1:
        return arr
    
    pivot = arr[0]

    left = []
    right = []

    for x in arr [1:]:
        if x <= pivot:
            left.append(x)
        else:
            right.append(x)
    return quick_sort(left) + [pivot] + quick_sort(right)

arr = [8,3,1,7,0,10,2]
print(quick_sort(arr))

```
