#What is Two Pointers?

Use two indexes (pointers) instead of one loop or nested loops.

Usually:

One pointer starts from beginning

One pointer starts from end

# Reverse an array (Two Pointers)

```python
arr = [1, 2, 3, 4, 5]

left = 0
right = len(arr) - 1

while left < right:
    arr[left], arr[right] = arr[right], arr[left]
    left += 1
    right -= 1
print(arr)    
```

# Check Palindrome 

```python
s = "level"
left = 0
right = len(s) - 1
while left < right:
    if s[left] != s[right]:
        print(" not palindrome") 
        break 
    left += 1
    right -= 1
else:   
    print(" palindrome")
    
```

#Problem: Check if this string is a palindrome

```python
s = "radar"

left = 0
right = len(s) - 1
while left < right:
    if s[left] != s[right]:
        print("if string is not palindrome")
        break
    left += 1
    right -= 1
else:
    print("if string is palindrome")
```

#Reverse a string

```python
s = "python"

# convert string to list
arr = list(s)

left = 0
right = len(s) - 1
while left < right:
    arr[left], arr[right] = arr[right], arr[left]      
    left += 1
    right -= 1
s = "".join(arr)    
print(s)
```

#Check if array is sorted

```python
arr = [1, 2, 3, 4, 5]

is_sorted = True

for i in range(len(arr)- 1):
    if arr[i]>arr[i+1]:
        is_sorted = False
        break
if is_sorted:
    print("sorted")
else:
    print("not sorted")    
```

```python
s = "banana"
count = 0

for char in s:
    if char == 'a':
        count += 1
print(count)
```

```python
arr = [1, 2, 3, 4, 5]

freq = {}
for num in arr:
    if num in freq:
        freq[num] += 1
    else:
        freq[num] = 1
print(freq)
```
