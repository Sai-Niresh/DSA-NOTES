```python
# Find the student with highest and lowest score
def student_high_low(scores):
    max_val = scores[0]
    min_val = scores[0]
    for num in scores:
        if num > max_val:
            max_val = num
        if num < min_val:
            min_val = num
    return max_val, min_val

scores = [85, 92, 78, 90, 88]
high, low = student_high_low(scores)
print(f"Highest score: {high}, Lowest score: {low}")
```

```python
# Does any pair add up to 16? Try it!

def has_pair(nums, target):
    left = 0
    right = len(nums) - 1
    while left < right:
        current_sum = nums[left] + nums[right]
        if current_sum == target:
            return True
        elif current_sum < target:
            left += 1
        else:
            right -= 1
    return False    
nums = [2, 4, 6, 8, 10, 12]
print(has_pair(nums, 16))

```

```python
# What is the maximum sum of 3 consecutive numbers

def max_sum_(nums,k):
    window_sum = sum(nums[:k])
    max_sum = window_sum
    for i in range(k, len(nums)):
        window_sum += nums[i]
        window_sum -= nums[i-k]
        max_sum = max(max_sum, window_sum)
    return max_sum
nums = [3, 5, 2, 8, 1, 4, 7]
k = 3
print(max_sum_(nums, k))
```

```python
# Find the first duplicate in a list
# Count frequency of [4, 5, 4, 6, 5, 4, 7]
def count_frequency(nums):
    freq = {}
    for num in nums:
        if num in freq:
            freq[num] += 1
        else:
            freq[num] = 1
    return freq
nums = [4, 5, 4, 6, 5, 4, 7]
print(count_frequency(nums))
```

```python
#Find first duplicate in [9, 3, 6, 3, 8, 9]
def first_duplicate(nums):
    seen = set()
    for num in nums:
        if num in seen:
            return num
        seen.add(num)
    return None
nums = [9, 3, 6, 3, 8, 9]
print(first_duplicate(nums))
```
