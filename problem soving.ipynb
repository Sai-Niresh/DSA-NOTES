```python
# two variable pattern
num = [5, 10, 20, 8, 15]
largest = float('-inf')
second_largest = float('-inf')
for x in num:
    if x > largest:
        second_largest = largest
        largest = x
    elif x > second_largest and x != largest:
        second_largest = x
if second_largest == float('-inf'):
    print("no second largest number")
else:
    print(second_largest)
```

```python
#Problem 1 — Reverse a list using two pointers
arr = [10, 20, 30, 40, 50]
left = 0
right = len(arr) - 1
while left < right:
    arr[left], arr[right] = arr[right], arr[left]
    left += 1
    right -= 1
print(arr)
```

```python
# Check whether a string is a palindrome
text = "madam"
left = 0
right = len(text) - 1
while left < right:
    if text[left] != text[right]:
        print("not Palindrome")
        break
    left += 1
    right -= 1
else:
    print("palindrome")
```

```python
# sorting pattern buble 

arr = [4, 2, 7, 1]

for i in range(len(arr)):
    for j in range(len(arr)-i - 1):
        if arr[j] > arr[j+1]:
            arr[j], arr[j+1] = arr[j+1], arr[j]
print(arr)
```

```python
# selectio sort pattern
arr = [4, 2, 7, 1]

for i in range(len(arr)):
    min_index = i
    for j in range(i+1, len(arr)):
        if arr[j] < arr[min_index]:
             min_index = j
        arr[i], arr[min_index] = arr[min_index], arr[i]
print(arr)
```

```python
def find_smallest_int(arr):
    min_val = arr[0]
    for x in arr:
        if x < min_val:
            min_val = x
    return min_val
arr = [34, 15, 88, 2]

find_smallest_int(arr)
```

```python
def basic_op(operator, value1, value2):
    if operator == '+':
        return value1 + value2
    elif operator == '-':
        return value1 - value2
    elif operator == '*':
        return value1 * value2
    elif operator == '/':
        return value1 / value2
    else:
        raise ValueError("Unsupported operator")
```
