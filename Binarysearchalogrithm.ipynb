#What is Binary Search?

Binary Search finds an element in a SORTED array by cutting the search space in half.

⚠️ Very important rule:

Binary Search works ONLY on sorted arrays.

🧠 Simple idea (plain English)

Instead of checking every element:

Look at the middle

If middle = target → done

If target is smaller → search left half

If target is bigger → search right half

```python
arr = [1, 3, 5, 7, 9]
target = 7

low = 0
high = len(arr)-1

found = False

while low<=high:
    mid = (low + high)//2

    if arr[mid] == target:
        found = True
        break
    elif arr[mid] < target:
        low = mid + 1
    else:
        high = mid - 1

if found:
    print("Element found")

else:
    print("Element not found")                
```
