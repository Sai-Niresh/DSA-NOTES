#Topic: Linked List (Basics)

This is Pattern #14.

Step 1: What is a Linked List?

A Linked List is a collection of nodes connected together.

Each node contains:

[data | next]


Example:

10 → 20 → 30 → 40 → None


10 is first node (head)

40 points to None (end)

Step 2: Difference between Array and Linked List

Array:

[10, 20, 30, 40]


Stored continuously in memory.

Linked List:

10 → 20 → 30 → 40


Stored separately but connected using links.

Step 3: Node structure in Python

Each node has:

data

next pointer

Code:

class Node:
    def __init__(self, data):
        self.data = data
        self.next = None


This creates one node.

Step 4: Create nodes manually
node1 = Node(10)
node2 = Node(20)
node3 = Node(30)


Now connect them:

node1.next = node2
node2.next = node3


Structure becomes:

10 → 20 → 30 → None

Step 5: Traversing Linked List

Traversal pattern is same as traversal pattern you learned.

Code:

current = node1

while current:
    print(current.data)
    current = current.next


Output:

10
20
30

Step 6: Visual understanding

Memory view:

node1 → [10 | address of node2]
node2 → [20 | address of node3]
node3 → [30 | None]

Step 7: Core Linked List Pattern

Traversal formula:

current = head

while current:
    process(current.data)
    current = current.next


This is the most important Linked List pattern.

Step 8: When to use Linked List

Problems with:

insertion

deletion

dynamic size

memory efficient operations

Step 9: Pattern recognition keywords

If problem says:

linked list

node

next pointer

head

→ Use Linked List pattern

#Linked List Problem 1: Traverse and print all elements
Problem

You are given a linked list:


10 → 20 → 30 → 40 → None

```python


class Node:
   def __init__(self, data):
    self.data = data
    self.next = None

head = Node(10)
second = Node(20)
third = Node(30)
fourth = Node(40)

head.next =second
second.next = third
third.next = fourth

current = head

while current:
    print(current.data)
    current = current.next
print(current)    

```

#Linked List Problem 2: Search for a value
Problem

Given linked list:

10 → 20 → 30 → 40 → None


Search for value:

target = 30

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None

head = Node(10)
second = Node(20)
Third = Node(30)
Fourth = Node(40)

head.next = second
second.next = Third
Third.next = Fourth

target = 30

current = head
while current:
    if current.data == target:
        print("Target found")
        break
    current = current.next
else:
    print("Target not found")         
```

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None

head = Node(10)
Second = Node(20)
Third = Node(30)
Fourth = Node(40)

head.next = Second
Second.next = Third
Third.next = Fourth

count = 0

current = head
while current:
    count += 1
    current = current.next
print("Length of the linked list:", count)
    
```
