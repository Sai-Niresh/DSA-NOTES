📌 What is a Stack?

A Stack is a data structure that follows
LIFO → Last In, First Out

🧠 Simple real-life example

Think of a stack of plates 🍽️

You put a plate on top

You remove the top plate first

That’s exactly how a stack works.

🔑 Stack Rules

Only two main operations:

1️⃣ Push → add element
2️⃣ Pop → remove last element

You can only work with the TOP element.

```python
# reverse a string uing stack
s = "python"
stack = []
rev= ""

for ch in s:
    stack.append(ch)
while stack:
    rev += stack.pop()
print(rev)        

```

```python
s = "() ()"
stack = []
balanced = True
for ch in s:
    if ch == '(':
        stack.append(ch)
    elif ch == ')':
        if not stack:
            balanced =False
            break
        stack.pop()

if not stack:
    print("Balanced")
else:
    print("Not Balanecd")            
```

```python
s = "aabbccdd"
stack = []
for ch in s:
    if not stack or stack [-1]!= ch:
        stack.append(ch)


result = "".join(stack)
print(result)
```

```python
s = "{[()]}"
Stack = []
vaild = True

pairs = {')':'(','[':'[','{':'}'}

for ch in s:
    if ch == '{[(':
        Stack.append(ch)
    elif ch ==')]}':
        if not Stack or stack [-1] != pairs[ch]:
            vaild =False
            break
        stack.pop()
if vaild and not Stack:
    print("vaild")
else:
    print("Invaild")                
```

```python
a = ["abc", "xyz", "pqr","amn", "xpz"]
r = ["ab", "cb","rb"]

result = {}

for pattern in r:
    result[pattern] = set()

    for word in a:
        for ch in pattern:
            if ch in word:
                result[pattern].add(word)
                break

for key in result:
    print(f'"{key}" :{result[key]}')

```
