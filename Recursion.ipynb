#RECURSION (FROM SCRATCH)

Don’t worry — I’ll make this very simple.

📌 What is Recursion?

Recursion = A function calling itself

That’s the whole idea.

🧠 Real-life example

Think of climbing stairs:

To reach step 5, you first reach step 4
To reach step 4, you first reach step 3
To reach step 3, you first reach step 2
To reach step 2, you first reach step 1

This is recursion.

🔑 Every recursion has 2 parts
1️⃣ Base Case → STOP condition
2️⃣ Recursive Case → Function calls itself

Without base case → infinite loop ❌

Example 1: Print numbers from 1 to 5
Normal way (loop)

```python
for i in range(1, 6):
    print(i)
```

# Recursion way

```python
def print_numbers(n):
    if n == 0: # Base case
        return 
    
    print_numbers(n - 1) # Recursive call
    print(n)

print_numbers(5)    
```

#Problem: Sum of numbers from 1 to n using recursion

```python
def sum_numbers(n):
    if n == 0:
        return 0
    return n + sum_numbers(n - 1)

print(sum_numbers(5))    
```

#Problem: Sum of numbers from 1 to n using recursion

```python
def factorial(n):
    if n == 0:
        return 1
    return n * factorial(n - 1)
print(factorial(5))
  
```
