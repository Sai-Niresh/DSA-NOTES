DAY 2 — SEARCHING ALGORITHMS

Searching means finding a specific element in a list (array).

🟩 1. What is Searching?

When you look for something in a group of items,
you’re doing a search.

Example (real life):
You look for your friend’s name in your contacts list → that’s a search.

We have two main types of searching:

Type	When to Use	Speed
Linear Search	For unsorted data	Slower
Binary Search	For sorted data	Faster
🔵 2. LINEAR SEARCH (Simple Search)

You go through the list one by one until you find the element.

Example

Find 30 in [10, 20, 30, 40, 50].

Step-by-step logic

Start from the first element.

Check if it equals the target (30).

If yes → found!

If no → move to the next element.

Repeat until the end.

```python
arr = [10, 20, 30, 40, 50]
target = 30 

found = False
for x in arr:
    if x == target:
        found = True
        break 
if found:
    print("Found")
else:
    print("Not Found")        
```

BINARY SEARCH (Fast Search)

Used when the list is sorted (in ascending or descending order).

Instead of checking every element,
we divide the list into halves again and again — like guessing a number game.

Real-life example

Imagine you’re searching for a word in a dictionary:

You open the middle → check if your word comes before or after → go to that half.

You keep cutting the list in half until you find it.

That’s binary search.


```python
arr = [10, 20, 30, 40, 50, 60, 70]
target = 50

low = 0
high = len(arr)-1
found = False


while low<= high:
    mid = (low + high) // 2
    if arr[mid] == target:
        found = True
        break 
    elif arr[mid] < target:
        low = mid + 1
    else:
        high = mid - 1
if found:
    print("Element found!")
else:
    print("Element")                
 
```

2. Bubble Sort (First Sorting Algorithm)

Bubble sort compares two adjacent elements and swaps them if they are in the wrong order.

🔥 Real-life example:

Imagine shaking a soda bottle → big bubbles rise to the top.

Similarly in bubble sort,
the biggest number rises to the end of the list first.

```python
arr = [5, 1, 4, 2, 3]

for i in range(len(arr)):
    for j in range(len(arr)- 1):
        if arr[j] > arr[j + 1]:
            arr[j], arr[j+1] = arr[j+1], arr[j]
print(arr)            
```

```python
# Reverse an array (Two Pointers)
arr = [1, 2, 3, 4, 5]
left = 0
right = len(arr) - 1
while left < right:
    arr[left], arr[right] = arr[right], arr[left]
    left += 1
    right -= 1
print(arr)
```

```python
# check palindrome
s = "madam"
left = 0
right = len(s) - 1
while left < right:
    if s[left] != s[right]:
        print("Not Palindrome")
        break   
    left += 1
    right -= 1
else:
    print("Palindrome")
```
